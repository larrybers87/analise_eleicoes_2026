# 13 · Concentração dos votos entre municípios

Candidatos com **mais de 1% dos válidos nacionais** (regra do projeto; os cálculos usam os 5.571 municípios brasileiros).

- **Gini de votos absolutos**: concentração dos votos do candidato entre municípios. Reflete também o tamanho dos municípios.
- **Gini de % sobre válidos**: dispersão do apoio entre municípios, cada um com peso 1.
- **Nº de municípios para 50%**: quantos municípios (do maior para o menor em votos) somam metade dos votos do candidato.

**Ressalva**: concentração entre municípios é propriedade da distribuição geográfica agregada. Não diz quem votou em quem dentro de cada município.

```python
import sys, pathlib

sys.path.insert(0, str(pathlib.Path("..") / "src"))
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

pd.set_option("display.width", 200)
pd.set_option("display.max_columns", 30)
from eleicao.analise import composicao as cp, base, carga, graficos as g
from eleicao.analise import (
    participacao as pa,
    concentracao as cc,
    swing as sw,
    perfil,
    espacial as esp,
    bolsoes as bo,
    mapa_mente as mm,
)
```

```python
b = cp.base_2026()
nac = cp.percentuais_nacionais(cp.candidatos_2026_br(), base.apenas_brasil(b))
nrs = [nr for nr, p in sorted(nac.items(), key=lambda kv: -kv[1]) if p > 1.0]
print("candidatos > 1% nacional:", nrs)
tab = cp.tabela_concentracao(b, nrs)
tab["nome"] = tab.nr_candidato.map(g.NOME_CURTO)
print(tab.round(4).to_string(index=False))
```

```python
curvas = cp.lorenz_por_candidato(b, nrs)
fig = g.lorenz_candidatos(
    curvas, "Curva de Lorenz dos votos por município (candidatos > 1% nacional)"
)
g.salvar(fig, "concentracao_lorenz")
plt.show()
```
